## Introduction

Until today, every picture we made showed **one** variable at a time. Today we look at **two together**. A **scatter plot** draws one dot for each person (or each run of a model), with one variable going across and the other going up.

The **correlation**, written ***r***, is one number that measures how closely the dots in a scatter plot follow **one straight line**:

- *r* is always between **-1** and **1**.
- If the dots go **up** to the right, *r* is **positive**. If they go **down** to the right, *r* is **negative**.
- Dots close to a line make a **strong** correlation: *r* near 1 or -1. Dots far from any straight line make a **weak** correlation: *r* near 0.

Today you will push *r* around and see what it takes to make it big or small.

The cell below puts the two tables into Python, and draws each one with `plt.scatter()`. It works like `plt.plot()` from class 6.1, but it draws one dot for each row and does not join them with a line:

````python
table_1 = pd.DataFrame({
    "x": [-3, -2, -1, 0, 1, 2, 3],
    "y": [2, 2.5, 3, 3.5, 4, 4.5, 5]
})
table_2 = pd.DataFrame({
    "x": [-4, -3, -2, -1, 0, 1, 2, 3, 4],
    "y": [18, 11, 6, 3, 2, 3, 6, 11, 18]
})

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)                          # 1 row, 2 columns: draw in place 1
plt.scatter(table_1["x"], table_1["y"])       # one dot per row: x across, y up
plt.xlabel("x")
plt.ylabel("y")
plt.title("Table 1")

plt.subplot(1, 2, 2)                          # 1 row, 2 columns: draw in place 2
plt.scatter(table_2["x"], table_2["y"])
plt.xlabel("x")
plt.ylabel("y")
plt.title("Table 2")

plt.tight_layout()
plt.show()
````

### Setting up and running albedo-sweep

The second sweep changes **albedo**: how much of the sunlight that reaches the ground is reflected straight back into space. It goes from 0 to 1:

- **0** means the ground reflects none of the sunlight and absorbs all of it, like black rock.
- **1** means the ground reflects all of the sunlight, like a perfect mirror.

On the real Earth, fresh snow and ice have a high albedo, around 0.8, and dark ocean water a low one, under 0.1. In the model, the slider starts at 0.6, and you can watch it work: yellow sunlight hits the ground, and some of it bounces back up while the rest turns into red heat. This sweep looks only at high albedo, from **0.8** to **1**, in small steps of **0.01**.
