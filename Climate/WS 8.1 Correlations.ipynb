# Part 1: Stretching the Correlation

Recall how we described the correlation in class:

The correlation coefficient is a statistical measure that describes the extent to which two variables change in a linear way together.

- The value of *r* (or rho, $\rho$) ranges smoothly from **-1.0 to 1.0**.
- If one variable tends to go up when the other goes up:
    - There is a **positive correlation**, $0 < r \leq 1$
    - $r = 1$ implies a perfect positive correlation (an increasing line).
- If one variable tends to go down when the other goes up:
    - There is a **negative correlation**, $-1 \leq r < 0$
    - $r = -1$ implies a perfect negative correlation (a decreasing line).
- When $r = 0$, the variables are not linearly correlated.

Correlations not only differ in their directions, but also their strength.

- Data that are more spread out have a **weaker correlation** (*r* near 0)
- Data that are closer to a line have a **stronger correlation** (*r* near 1 or -1)
