# Worksheet 5.1: Agent-Based Models and Emergence

## Introduction

In today's class, we explore **agent-based models (ABMs)**—simulations where individual "agents" follow simple rules, and we observe what happens when many agents interact simultaneously.

A key concept: **emergent behavior**. Some systems are *resultant*—the outcome is a direct, predictable consequence of the inputs (like a calculator). But in emergent systems, the whole is more than the sum of its parts. Simple rules can produce complex, surprising patterns that aren't obvious from the rules themselves.

Why does this matter? Because many real-world systems—from social dynamics to climate—exhibit emergence. And when systems are emergent rather than resultant, we can't just "figure them out" by thinking hard. We need models to explore them, and **quantitative tools** to make sense of what we observe.
