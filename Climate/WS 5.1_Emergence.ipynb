## Introduction

You just played the Heroes and Cowards game. Today we look at **agent-based models (ABMs)**: computer simulations where many individual "agents" each follow simple rules, and we watch what happens when they all act at once.

A key idea for today is **emergence**. Some systems are *resultant*: the outcome is a direct, predictable result of the inputs, like a calculator. Press 2 + 2 and you get 4, every time, and you can explain why. In an *emergent* system, the whole is more than the sum of its parts. Simple rules can produce complex, surprising patterns that you could not guess from the rules themselves.

Why does this matter? Many real systems are emergent, from crowds of people to the climate. When a system is emergent, we cannot just figure it out by thinking hard. We need models to explore it, and **quantitative tools** to make sense of what we see.

### Key takeaways

- **Emergent patterns belong to the whole group, not to any one agent.** You cannot find them by asking one person what they were doing.
- **Agent-based models let us explore emergence** by running the rules many times.
- **The same model with the same settings can give different results** when chance is part of it.
- **So one run is not enough.** When results vary, we need to say two things about them: where the **middle** is, and how **spread out** they are. Those are the next questions in this unit, starting next class.
