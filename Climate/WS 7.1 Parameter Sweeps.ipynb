### The mean at each value

We have 10 runs at each of 5 values, all mixed together in one table. We want the mean temperature at **each** value. pandas has a tool for this called `.groupby()`. It splits the table into groups, one for each value of `sun-brightness`, and then works something out for every group at once.

Read the line below from left to right: take `sweep`, **group** it by `sun-brightness`, take the `temperature` column, and find the **mean** of each group. We store the answer in a variable called `means`, and then print it:

````python
means = sweep.groupby("sun-brightness")["temperature"].mean()
print(means)
````

You can also ask for both at once, with `.agg()` (short for "aggregate", which means "summarize"). This makes a small table with a `mean` column and a `std` column. We will use this table for the plots in Part D:

````python
summary = sweep.groupby("sun-brightness")["temperature"].agg(["mean", "std"])
print(summary)
````

## Part D: Plotting the sweep

A table of numbers is hard to read at a glance. A **line chart** is easier. The code below draws one straight from `summary`, the table you made in Part C, so there are no numbers to type in.

- `plt.plot()` draws a line through points. First it takes what goes **along the bottom**, then what goes **up the side**.
- Along the bottom: `summary.index`, the values of `sun-brightness`. (These are the row labels on the left of `summary`; Python calls them the **index**.)
- Up the side: `summary["mean"]`, the mean temperature at each value.
- `marker="o"` puts a dot on each point. The next three lines add labels, and `plt.show()` draws the chart.

Run the cell:

````python
plt.plot(summary.index, summary["mean"], marker="o")
plt.xlabel("Sun brightness")
plt.ylabel("Temperature at the end of the run")
plt.title("Sun brightness sweep")
plt.show()
````

A line through the means hides something: each mean comes from 10 runs that bounced around. `plt.errorbar()` draws the same line, plus a bar above and below each point to show the bounce. It needs one more thing than `plt.plot()`: `yerr=`, which says how long to make the bars. We want each bar to be one standard deviation long, and the standard deviations are in `summary["std"]`.
