**Python has a command for the mean.** Adding up nine numbers by hand gets old fast. **NumPy** is a Python library full of tools for working with numbers, and one of them works out the mean for you. Put the numbers inside square brackets, separated by commas. Type a negative number with a minus sign, like `-1`:

```python

```

The first line loads NumPy and gives it the short name `np`. The second line works out the mean of −1, 2 and 5, and running it would print `2.0`. You will see a lot more of NumPy next week.
