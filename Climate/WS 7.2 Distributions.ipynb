## Introduction

At the start of class we looked at **histograms** of the waits between eruptions of Old Faithful. A histogram sorts numbers into **bins**, ranges of values, and the height of each bar shows how many values landed in that range. With 3 bins the two humps disappeared, and with 100 bins the picture turned into spiky noise: the number of bins changes what you see. The whole picture, every value and where it sits, is called a **distribution**.

So far we have described a pile of numbers with two numbers: the **mean** for the center (class 5.2) and the **standard deviation** for the spread (class 6.2). Today's question is what those two numbers **miss**.

The reading for today gave us words for describing a distribution:

- **Shape:** is it **symmetric** (the same on both sides, like a bell), or **skewed** (a long tail on one side)? A distribution with a long tail on the right is **skewed to the right**.
- **Location:** where is a typical value?
- **Variation:** how spread out are the values?
- **Outliers:** are there values far away from all the others?

Each row is one student: their `Class` (A to I) and their `Score`. We will look at the classes three at a time.

A histogram is the **result** of something that happened to real people. A good description says **what** the distribution looks like (its location, variation and shape, from the reading) and also **why** it might have come out that way.

### Picking out one class

The table has all 450 students mixed together. To draw a histogram of Class A, we need **only** Class A's scores. That takes three small steps.

**Step 1: ask a question about every row.** Run this:

````python
exams["Class"] == "A"
````

`exams["Class"]` is the `Class` column: a column name inside square brackets picks out that column, just like `results["temperature"]` in class 6.2. Then `==` asks "is this equal to?" (two equals signs, because one equals sign already means "store this in a variable"). The answer is one `True` or `False` for every row: `True` for each student in Class A.

**Step 2: keep only the `True` rows.** When you put that question inside square brackets after a table, pandas keeps the rows where the answer is `True` and throws away the rest, like a sieve:

````python
class_a_rows = exams[exams["Class"] == "A"]
class_a_rows.head()
````

**Step 3: take the column you want.** `class_a_rows` still has both columns. We only want the scores:

````python
class_a = class_a_rows["Score"]
class_a.head()
````

### Classes A, B and C

First, pick out Class B and Class C the same way:

````python
class_b_rows = exams[exams["Class"] == "B"]
class_b = class_b_rows["Score"]
class_c_rows = exams[exams["Class"] == "C"]
class_c = class_c_rows["Score"]
````

Now draw the three histograms, one above the other. `plt.subplot(3, 1, 1)` means "a grid of **3** rows and **1** column, and draw in place number **1**", the top. Each histogram gets the same `range=`, so the scales line up and we can compare them by eye.

````python
plt.figure(figsize=(8, 7))                    # one figure, 8 wide and 7 tall

plt.subplot(3, 1, 1)                          # 3 rows, 1 column: draw in place 1
plt.hist(class_a, bins=13, range=(35, 100))   # the same range for every class
plt.ylabel("Number of students")
plt.title("Class A")

plt.subplot(3, 1, 2)                          # 3 rows, 1 column: draw in place 2
plt.hist(class_b, bins=13, range=(35, 100))
plt.ylabel("Number of students")
plt.title("Class B")

plt.subplot(3, 1, 3)                          # 3 rows, 1 column: draw in place 3
plt.hist(class_c, bins=13, range=(35, 100))
plt.ylabel("Number of students")
plt.xlabel("Score")
plt.title("Class C")

plt.tight_layout()
plt.show()
````

### Key takeaways

- **Always look at the picture.** Location, variation and shape belong together. The mean sees location and the SD sees variation, but only the picture shows all three: Classes G, H and I have the same mean and the same SD and still look different.
- **A shape has a cause.** A wall on one side, where values cannot go any further, often leaves a long tail on the other side.
- **A list of values in BehaviorSpace:** `["SLIDER-NAME" VALUE VALUE VALUE]`, with no inner brackets. A range has them: `["SLIDER-NAME" [START STEP END]]`.
- **Picking out one group:** `ROWS = TABLE[TABLE["COLUMN"] == VALUE]` keeps the rows where the answer is `True`, and `ROWS["OUTCOME"]` takes one column from them. Then `plt.hist()` each group with the same `range=`, stacked with `plt.subplot()`.
