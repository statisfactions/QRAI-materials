## Part B: Your First Histogram

A **histogram** groups data into bins (ranges) and shows how many values fall in each bin. Unlike a bar chart (which shows categories), a histogram shows the *distribution* of numerical data -- where values cluster, how spread out they are, and whether the shape is symmetric or skewed.

In matplotlib, we use `plt.hist()`. Run the cell below to create a histogram of Class A exam scores.

````python
plt.hist(class_a["Score"])
plt.xlabel("Score")
plt.ylabel("Count")
plt.title("Class A Exam Scores")
plt.show()
````

Each bar represents a **bin** -- a range of scores. The height of each bar shows how many students scored within that range. By default, `plt.hist()` uses 10 bins.

### Comparing Plots with `plt.subplot()`

It's hard to compare histograms when you can only see one at a time. `plt.subplot()` lets you put multiple plots in one figure.

`plt.subplot(rows, cols, n)` means: create a grid with this many *rows* and *cols*, and draw in position *n*. For example, `plt.subplot(2, 1, 1)` means "2 rows, 1 column, position 1 (top)." Run the cell below to see Class A scores with two different bin values stacked vertically.

````python
plt.figure(figsize=(8, 6))

plt.subplot(2, 1, 1)
plt.hist(class_a["Score"], bins=5)
plt.ylabel("Count")
plt.title("Class A (bins=5)")

plt.subplot(2, 1, 2)
plt.hist(class_a["Score"], bins=20)
plt.ylabel("Count")
plt.xlabel("Score")
plt.title("Class A (bins=20)")

plt.tight_layout()
plt.show()
````

Let's visualize these three classes. When comparing histograms, setting the same `range` for all of them ensures the x-axes line up, making differences easier to spot. The `range` parameter in `plt.hist()` sets the minimum and maximum values for the x-axis. Run the cell below.

````python
class_a = data[data["Class"] == "A"]
class_b = data[data["Class"] == "B"]
class_c = data[data["Class"] == "C"]

plt.figure(figsize=(8, 8))

plt.subplot(3, 1, 1)
plt.hist(class_a["Score"], bins=10, range=(50, 72))
plt.ylabel("Count")
plt.title("Class A")

plt.subplot(3, 1, 2)
plt.hist(class_b["Score"], bins=10, range=(50, 72))
plt.ylabel("Count")
plt.title("Class B")

plt.subplot(3, 1, 3)
plt.hist(class_c["Score"], bins=10, range=(50, 72))
plt.ylabel("Count")
plt.xlabel("Score")
plt.title("Class C")

plt.tight_layout()
plt.show()
````
