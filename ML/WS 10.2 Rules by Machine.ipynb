To learn a rule, the computer needs two things:

- **`X`**: the features it may use. We list their names in `features`, then pick out those columns.
- **`y`**: the label it is trying to predict, `xhealth`.

We leave `xhealth` out of `X`, because that is the answer.

````python
features = ["eyes_orange", "hair_pink", "antennae_7", "weight", "height"]
X = dengueso_df[features]
y = dengueso_df["xhealth"]
````

Now the learning. This code comes from a Python library called **scikit-learn** (`sklearn`), which began in 2007 as a student's summer project and is now one of the most used machine-learning tools in the world.

- `DecisionTreeClassifier(max_depth=1)` makes an empty tree that may ask **one** question.
- `model.fit(X, y)` is the **training**: the computer tries every question and keeps the best one.
- `plot_tree(...)` draws the tree the way sklearn draws it.

````python
model = DecisionTreeClassifier(max_depth=1)
model.fit(X, y)

plt.figure(figsize=(10, 3))
plot_tree(model, feature_names=features, class_names=["sick", "well"], filled=True)
plt.show()
````

The top line of the top box is the question, written as `weight <= ...`: aliens for which it is **true** go **left**. Don't worry about the other lines (`gini`, `samples`, `value`).

The next cell draws the same tree the way Tuesday's tool did, with the training counts in each box.

````python
show_tree(model, X, y)
````

### Which mistakes?

"Right: 129 of 150" tells you how many mistakes, but not which kind. For a disease, there are two very different ways to be wrong: telling a sick alien it is well, or telling a well alien it is sick.

The next cell asks the tree for its prediction for every alien (`model.predict(X)`), then counts each combination of what the tree said and what was true. This table is called a **confusion matrix**.

````python
predictions = model.predict(X)
pd.crosstab(y, predictions, rownames=["really"], colnames=["tree says"])
````

Each of the four boxes has a name:

| | Tree says **sick** | Tree says **well** |
|---|---|---|
| **Really sick** | **True positive (TP)**: caught | **False negative (FN)**: missed |
| **Really well** | **False positive (FP)**: false alarm | **True negative (TN)**: correctly cleared |
