### How `DecisionTreeClassifier` works

`DecisionTreeClassifier` is a Python tool (from a library called `sklearn`) that builds a decision tree automatically from labeled data — the same kind of tree you built by hand in Arbor.

A decision tree has two kinds of boxes:

- **Nodes** — question boxes inside the tree. Each node asks a question about one feature (for example: *"Is this feature above or below some threshold?"*).
- **Leaves** — boxes at the bottom of the tree. Each leaf shows the model's final prediction: *sick* or *well*.

`max_depth` controls how many levels of questions the tree is allowed to ask. With `max_depth=1`, the tree can ask exactly **one question** — one node and two leaves:

```
      [feature ≤ ???]
       /             \
    YES               NO
 [sick leaf]     [well leaf]
```

To train the model we need two inputs:

- **`X`** — the features we give the model to work with. We store the column names in a list called `features`, then select those columns with `dengueso_df[features]`. Indexing a DataFrame with a list of names gives back a smaller DataFrame — the same structure you've been using with pandas, just fewer columns.
- **`y`** — the labels the model is trying to predict. Indexing with a single name (`'xhealth'`) gives back one column.

We can't pass the entire dataset to `X` because it contains `xhealth` — giving the model the answer during training would be cheating! We also leave out `eyes`, `hair`, and `antennae` since those are text columns and `sklearn` requires numbers.

Run the cell below.

````python
features = ['eyes_orange', 'hair_pink', 'antennae_7', 'weight', 'height']
X = dengueso_df[features]
y = dengueso_df['xhealth']
````

Now train the model and draw the tree.

- `DecisionTreeClassifier(max_depth=1)` creates a new, untrained tree that is allowed to ask one question.
- `model.fit(X, y)` is **training**: the algorithm searches through all possible questions on all 5 features and finds the one that best separates *sick* from *well* — similar to what Arbor does when you click *Build*.
- `plot_tree(...)` draws the tree. `feature_names=features` labels each node with the column name; `class_names=['sick', 'well']` labels each leaf; `filled=True` colors the nodes (orange tones = *sick* majority, blue = *well* majority).

Run the cell.

````python
model = DecisionTreeClassifier(max_depth=1)
model.fit(X, y)

plt.figure(figsize=(10, 3))
plot_tree(model, feature_names=features, class_names=['sick', 'well'], filled=True)
plt.title('Decision tree for dengueso')
plt.show()
````

> **Reading the tree diagram:** Each box shows four pieces of information:
>
> - **The question** (top line of each internal node): the feature and threshold the tree is testing. Patients on the **left branch** satisfy the condition (≤ threshold); patients on the **right branch** do not (> threshold). Boxes at the very bottom — the **leaves** — show no question, only a prediction.
> - **gini**: how mixed the patients in that box are. 0 means everyone is the same class (pure — easy to predict); 0.5 means a perfect 50/50 split (maximally mixed — hardest to predict). Lower gini is better.
> - **samples**: the number of training patients in that box.
> - **value**: `[a, b]` — patient counts by class, **in the order of `class_names`**. Because we wrote `class_names=['sick', 'well']`, **value = [number sick, number well]**. The majority class in each leaf becomes its prediction (shown at the bottom).
> - **Color**: orange tones mean *sick* is the majority; blue tones mean *well* is the majority. Deeper color = more lopsided split.

Next, let's measure accuracy. `model.predict(X)` runs every row in `X` through the trained tree and returns a prediction for each patient. Run the cell below to see what the predictions look like.

````python
predictions = model.predict(X)
print(predictions[:10])   # show the first 10 predictions
````

Each entry is either `'sick'` or `'well'` — one prediction per patient. Now compare each prediction to the true label in `y`. The expression `predictions == y` checks each pair: it gives `True` where they match and `False` where they don't. Because Python treats `True` as 1 and `False` as 0, `.sum()` counts the number of correct predictions. Run the cell.

````python
is_correct = (predictions == y)
print(is_correct[:10])   # True = correct, False = wrong

n_correct = is_correct.sum()   # True counts as 1, False as 0
print(n_correct, 'out of', len(y), 'correct')
````

### The confusion matrix

Accuracy tells you how often the model is right overall. But *which* mistakes it makes matters too — missing a sick patient is very different from wrongly flagging a healthy one.

Run the cell below. It builds a table comparing what the model predicted against what was actually true.

````python
cm = pd.crosstab(y, predictions3, rownames=['actual'], colnames=['predicted'])
cm
````

The table has four cells. Each cell counts a different combination of actual and predicted label. These four combinations have standard names:

| | Predicted **sick** | Predicted **well** |
|---|---|---|
| **Actually sick** | **True Positive (TP)** — caught correctly | **False Negative (FN)** — missed |
| **Actually well** | **False Positive (FP)** — false alarm | **True Negative (TN)** — correctly cleared |

- **TP**: model said *sick*, patient is actually *sick* — correct.
- **TN**: model said *well*, patient is actually *well* — correct.
- **FP**: model said *sick*, patient is actually *well* — a false alarm.
- **FN**: model said *well*, patient is actually *sick* — a missed case.
