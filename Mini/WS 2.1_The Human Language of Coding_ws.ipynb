## `print()` as a function

Python uses *functions* to get things done. A function is a command to perform a specific task. A function always begins with the function name, and then parentheses with any arguments inside.

Consider the code we ran above:

```python
print("Hello World")
```

Here's a breakdown of the elements of a function in Python, with this command as an example:

* **Function Name:** This is the name you use to call the function. In our example, the function name is `print`.

* **Arguments:** These are the values you pass into the function when you call it. They always come inside the parentheses after the function name. Here, we pass the text "Hello World" to the function.

## Assigning variables

In Python, variable assignment is like giving a name to a value. You use the `=` symbol to associate a variable name (on the left) with a specific value (on the right). This allows you to store and use data easily throughout your program.

**Example:**

````python
message = "Hello, world!"  # Assign the string "Hello, world!" to the variable 'message'
print(message)             # Display the value stored in 'message'
````

Look closely at the quotation marks. When we made the variable, `"Hello, world!"` had quotation marks around it. When we printed it, `message` did not.

## What are Libraries?

All the functions we've used so far, like `print()` and `round()`, come built into Python. But here's what makes Python truly powerful: **anyone can write additional functions and share them with the world.**

Developers package related functions together into a **library** (sometimes called a "package" or "module"). Think of it like this:

- Python itself is like a basic toolkit with essential tools (hammer, screwdriver, measuring tape)
- Libraries are like specialized toolkits you can add: a plumbing kit, an electrical kit, a woodworking kit...

There are libraries for almost anything you can imagine: analyzing data, creating websites, making games, generating art, and, most relevant for us today, **communicating with AI**.

## Importing a library

Before you can use a library, you need to **import** it, which is basically telling Python "hey, load this toolkit so I can use it."

```python
from google.colab import ai
```

This says: "From the `google.colab` library, just grab the `ai` tool." It's like saying "I don't need the whole electrical toolkit, just hand me the voltage tester."

Run the cell below. Nothing will seem to happen, and that's normal: importing just gets the toolkit ready.

````python
from google.colab import ai
````

## Talking to the AI

Here's the moment of truth! Run the cell below:

````python
answer = ai.generate_text("What is Quantitative Reasoning? Respond in two sentences.")
print(answer)
````

Let's break it down:

| Code | What it does |
|------|--------------|
| `ai.generate_text(...)` | Sends your text to the AI and gives back its reply. The dot means "the `generate_text` function, from inside the `ai` toolkit." |
| `answer = ...` | Stores the reply in a variable called `answer`, just like in Part B |
| `print(answer)` | Shows the reply on the screen |
