## A.1 `print()` as a function
Python uses *functions* to get things done. A function is a command to perform a specific task. A function always begins with the function name, and then parenthesis with any arguments inside.

Consider the code we ran above:

````
print("Hello World")
````

Here's a breakdown of the elements of a function in Python, with this command as an example:

* **Function Name:** This is the unique identifier you use to call or execute the function. In our example, the function name is `print`.

* **Arguments:** These are the values you pass into the function when you call it. Here, we pass the text "Hello World" to the function.

* **Return Value (this function doesn't have one):**  Some functions produce a result or value that can be used elsewhere in your program.  The `print` function doesn't explicitly return a value, but it performs the action of displaying output.

**In summary, the "print" function demonstrates these core elements:**

* It has a name (`print`) that we use to invoke it.
* It takes parameters (the objects to print, and optional formatting parameters) that control its behavior.  These always come in parentheses after the function.
* While it doesn't explicitly return a value, it fulfills its purpose by producing output

## A.2 Assigning variables

In Python, variable assignment is like giving a name to a value.  You use the `=` symbol to associate a variable name (on the left) with a specific value (on the right).  This allows you to store and manipulate data easily throughout your program.

**Example:**

````
message = "Hello, world!"  # Assign the string "Hello, world!" to the variable 'message'
print(message)             # Display the value stored in 'message'
````

**Assigning Variables to Other Variables**

You can also assign the value of one variable to another. This creates a copy of the value, not a link to the original variable.

````
original_message = "Hello, world!"
new_message = original_message
new_message = "Greetings!"
print(original_message)  # Output: Hello, world!
print(new_message)       # Output: Greetings!
````

# Part C: Libraries


## C.1 What are Libraries?



All the functions we've used so far — like `print()` and `len()` — come built into Python. But here's what makes Python truly powerful: **anyone can write additional functions and share them with the world.**

Developers package related functions together into a **library** (sometimes called a "package" or "module"). Think of it like this:

- Python itself is like a basic toolkit with essential tools (hammer, screwdriver, measuring tape)
- Libraries are like specialized toolkits you can add: a plumbing kit, an electrical kit, a woodworking kit...

There are libraries for almost anything you can imagine: analyzing data, creating websites, making games, generating art, and — most relevant for us — **communicating with AI**.

Today we'll set up the library that lets us talk directly to Google's Gemini AI from our code. This is the foundation for our temperature experiments in the next class!

## C.2 Importing Libraries

Before you can use a library, you need to **import** it — basically telling Python "hey, load this toolkit so I can use it."

There are two common ways to import:

### Pattern 1: Import with a nickname

```python
import math as m
```

This says: "Import the `math` library, but let me call it `m` for short."

Why use a nickname? Because typing `math` every time would get tedious! The nickname `np` is shorter and easier.

### Pattern 2: Import just one specific thing

```python
from google.colab import userdata
```

This says: "From the `google.colab` library, just grab the `userdata` tool."

This is like saying "I don't need the whole electrical toolkit — just hand me the voltage tester."

## C.4 Creating a Client

Now for the exciting part — let's create a connection to an AI!

Run the following cell:

````
from google import genai
from google.colab import userdata

client = genai.Client(api_key = userdata.get('GEMINI_API_KEY'))
````

Let's break down what each line does:

| Code | What it does |
|------|--------------|
| `from google import genai` | From the `google` library, import the `genai` tool |
| `from google.colab import userdata` | From the `google.colab` library, import the `userdata` tool |
| `userdata.get('GEMINI_API_KEY')` | Retrieve your secret key by its name |
| `genai.Client(api_key=...)` | Create a connection to Google's AI service |
| `client = ...` | Store that connection in a variable called `client` |

The `client` variable now holds something you can interact with. In programming, we call this an **object** — it's not just a number or text, but a "thing" that can do stuff.

### The Dot: Accessing What's Inside

You've now seen the dot (`.`) appear several times. In Python, the dot means "go inside this thing and access something."

Think of it like an address:
- `google.colab` means "inside `google`, find `colab`"
- `userdata.get(...)` means "inside `userdata`, use the `get` function"
- `genai.Client(...)` means "inside `genai`, use the `Client` function"

The dot lets you navigate into libraries, objects, and modules to find the tools you need. You'll use it constantly — it's one of the most common symbols in Python!

## C.5 Talking to the AI

Here's the moment of truth! To send a message to Gemini, we need to:
1. Go inside `client` to find `models`
2. Go inside `models` to find the `generate_content` function
3. Call that function with our message

This chain of dots looks like this:

````
response = client.models.generate_content(
    model='gemma-3n-e2b-it',
    contents="What is Quantitative Reasoning? Respond in two sentences."
)
print(response.text)
````

Let's trace through the dots:
- `client` — our connection to Google's AI service
- `client.models` — the part of the client that deals with AI models
- `client.models.generate_content(...)` — the function that sends a message and gets a response

And one more dot at the end:
- `response.text` — go inside the response object and get just the text part
